```python
import sys
sys.path.append('../')
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
# f_cnot = pd.read_csv('data/data_cnot_45_2A.txt')
# f_n2_20 = pd.read_csv('../cz/data/data_fig/data_cz_n2_20.txt')
f_theta = pd.read_csv('data/data_xgate_theta_truc44.txt')
f_theta_drag1 = pd.read_csv('data/data_xgate_theta_truc44_drag1.txt')
f_phi = pd.read_csv('data/data_xgate_phi_truc44.txt')
f_phi_drag1 = pd.read_csv('data/data_xgate_phi_truc44_drag1.txt')
```

```python
f_phi.to_numpy()
```

```python
fig, ax = plt.subplots(figsize=(6,3))
# plt.plot(f_cnot['tg'], 10**f_cnot['fidelity_qt'], '-', label='CNOT')   #r'$1-F_{qt}$')
# plt.plot(f_n2_20['tg'], 10**f_n2_20['fidelity_qt'], '-',  label=r'CZ')

# plt.plot(f_phi['tg'], 10**f_phi_drag1['fidelity_truc159'], '.-', label=r'$n_\phi$:truncation=159') 
plt.plot(f_phi['tg'], 10**f_phi['fidelity_truc44'], '.--', label=r'$n_\phi$:truncation=44')   
plt.plot(f_phi['tg'], 10**f_phi_drag1['fidelity_truc44'], '.--', label=r'$n_\phi$:truncation=44 (drag1)')   
# plt.plot(f_phi['tg'], 10**f_phi['fidelity_truc159'], '.-', label=r'$n_\phi$:truncation=159')   
# plt.plot(f_theta['tg'], 10**f_theta_drag1['fidelity_truc157'], '.-', label=r'$n_\theta$:truncation=157')
plt.plot(f_theta['tg'], 10**f_theta['fidelity_truc44'], '.--', label=r'$n_\theta$:truncation=44')
plt.plot(f_theta['tg'], 10**f_theta_drag1['fidelity_truc44'], '.--', label=r'$n_\theta$:truncation=44 (drag1)')

plt.yscale('log')
plt.xlabel('Gate time $t_g$ (ns)')
plt.ylabel('Error')
plt.title('Drag1: Use optimized pulse from 44-level')
plt.grid()

legend = plt.legend(    loc="best",    )
legend.get_frame().set_alpha(0.2)  # 0.5 makes the legend 50% transparent
# plt.savefig('cnot_fidelity_2A0_and_1A0.png',dpi=300,bbox_inches='tight')
```
