```python
def get_w_trans(truncation, evals, n_op, state_pop):
    thresh = 1e-4
    hspace = [0, 2]
    for s in hspace:
        for i in range(truncation):
            if np.abs(n_op[s, i] / (2 * np.pi)) > thresh and i not in hspace:
                hspace.append(i)
    hspace.sort()

    n_op = n_op[np.ix_(hspace, hspace)]
    w_trans = []
    trans_int = []
    n_trans = []
    for i, state_i in enumerate(hspace):
        if state_i in state_pop:
            for j, state_j in enumerate(hspace):
                if i < j and np.abs(n_op[i, j] / (2 * np.pi)) > thresh:
                    w_trans.append(np.round(np.abs(evals[state_i] - evals[state_j]), 8))
                    trans_int.append((state_i, state_j))
                    n_trans.append(np.round(np.abs(n_op[i, j] / (2 * np.pi)), 3))
    sort_idx = np.argsort(w_trans)
    w_trans = np.sort(np.round(w_trans, 3))
    n_trans = [np.round(n_trans[i], 3) for i in sort_idx]
    trans_int = [trans_int[i] for i in sort_idx]
    return w_trans, n_trans, trans_int
```

```python
def potential_zeropi(theta, phi):
    phiext = 0.0
    return -2 * EJ * np.cos(theta) * np.cos(phi - np.pi * phiext) + EL * phi**2 + 2 * EJ
```

```python
truc1 = 300
evals = eval0 - eval0[0]

# --- Arrow drawer ---
def draw_drive_arrow(bax, w_drive, label, dx_label=0.05, dy=0.1, color=colors[3], zorder=1):
    """
    Draw a vertical arrow at x=w_drive from y=0→y=dy, plus a nearby label.
    Uses `bax.annotate`/`bax.text`, so it works with Axes or BrokenAxes.

    Parameters
    - bax: Axes or BrokenAxes
    - w_drive: float (x in data units)
    - label: str (e.g., r"$\\omega_{d1}$")
    - dx_label: float, x-offset of label (data units)
    - dy: float, arrow height (data units)
    - color: color spec
    - zorder: draw order
    Note: `dy` is absolute (not auto-scaled). Ensure `w_drive` lies in a visible panel.
    """
    # Draw a vertical arrow from (w_drive, 0) up to (w_drive, dy)
    bax.annotate(
        "", xy=(w_drive, 0), xytext=(w_drive, dy),
        arrowprops=dict(color=color, shrink=0.02, headwidth=3, headlength=1.5, width=0.05, alpha=1),
        zorder=zorder,)
    # Label just above
    bax.text(w_drive + dx_label, dy * 1.15, label, color=color, ha="center", va="bottom", zorder=zorder + 1)



# Config & constants
alpha_other   = 1.0
medium_gray   = "0.75"
light_gray    = "0.9"
color_0       = colors[0]
color_2       = colors[1]

size_mark     = 2.5
size_text     = 5
width         = 0.8
linewidth_arrow = width

xlims = ((4.26, 5.1), (7.9, 9.3))   # broken x-axis ranges
thr   = 0.01                        # n_trans visibility threshold

# Classification sets
trans_lambda  = {(0, 9), (2, 9)}
trans_level_1 = {(0, 10), (2, 19), (9, 33), (9, 37), (9, 18)}
trans_level_3 = {(47, 69), (47, 70), (47, 91)}

# Per-class default styles
CLASS_STYLE = {
    "lambda": dict(color=None,        marker="s", vline="solid", alpha=1.0),
    "level1": dict(color="black",     marker="D", vline="--",    alpha=1.0),
    "level2": dict(color=medium_gray, marker=".", vline="--",    alpha=alpha_other),
    "level3": dict(color=light_gray,  marker="^", vline="--",    alpha=alpha_other),}

# Default label offsets (dx, dy)
DEFAULT_OFFSET = {
    "lambda": ( 0.02,  0.01),
    "level1": (-0.08, -0.11),
    "level2": (-0.03,  0.018),
    "level3": ( 0.02,  0.005),}

# Pair-specific offset overrides
OFFSET_OVERRIDE = {
    # lambda
    (0, 9): (0.015, -0.01),
    (2, 9): (0.12,   0.01),

    # level-1
    (0, 10): (-0.08, -0.11),
    (2, 19): ( 0.02, -0.11),
    (9, 18): (-0.08, -0.03),
    (9, 33): (-0.08, -0.11),
    (9, 37): (-0.08, -0.11),
    # (19,37): (-0.08, -0.11),
    # (10,37): ( 0.02, -0.11),

    # level-3
    (47, 69): (-0.08,  0.005),
    (47, 70): ( 0.02,  0.005),
    (47, 91): ( 0.02,  0.005),

    # level-2 examples
    (10, 25): ( 0.02,  0.005),
    (10, 42): ( 0.02,  0.005),
    (10, 23): ( 0.02,  0.005),
    (10, 40): ( 0.02, -0.05),
    (19, 37): (-0.08, -0.04),
    (19, 56): ( 0.02,  0.005),
    (33, 53): ( 0.02,  0.005),
    (33, 55): (-0.06,  0.005),
    (33, 69): (-0.07,  0.005),
    (33, 70): (-0.06,  0.005),
    # (47,70): ( 0.01,  0.01),
    # (9, 18): (-0.08, -0.04),
    # (9, 33): (-0.08, -0.04),
}

# Helper functions
def classify_pair(pair):
    if pair in trans_lambda:  return "lambda"
    if pair in trans_level_1: return "level1"
    if pair in trans_level_3: return "level3"
    return "level2"

def pair_style(pair):
    cls   = classify_pair(pair)
    style = CLASS_STYLE[cls].copy()
    if cls == "lambda":  # color depends on which lambda
        style["color"] = color_0 if pair == (0, 9) else color_2
    return cls, style

def label_offset(pair, cls):
    return OFFSET_OVERRIDE.get(pair, DEFAULT_OFFSET[cls])

def in_any_xlim(x, xlims):
    return any(lo < x < hi for lo, hi in xlims)

# Data
state_phi = [0, 2, 9, 10, 19, 33, 47]
w_trans, n_trans, trans_int = get_w_trans(truc1, evals, n_phi0, state_phi)

# Plot
fig = plt.figure(figsize=(3.4, 1.434))
bax = brokenaxes(xlims=xlims, wspace=0.05, d=0.004,despine=False, hspace=0.05)

for w, n, pair in zip(w_trans, n_trans, trans_int):
    if n <= thr or not in_any_xlim(w, xlims):
        continue

    cls, style = pair_style(pair)
    dx, dy     = label_offset(pair, cls)

    bax.scatter(w, n, color=style["color"], s=size_mark, marker=style["marker"], zorder=2, alpha=style["alpha"])
    ymax = n if style["vline"] == "solid" else max(n - 0.02, 0)
    bax.vlines(w, ymin=0, ymax=ymax,
               color=style["color"], linestyles=style["vline"],linewidth=width, zorder=1, alpha=style["alpha"])
    bax.text(w + dx, n + dy, f"{pair[0]}-{pair[1]}",
             rotation=90, fontsize=size_text,color=style["color"], alpha=style["alpha"])

# Special annotate for (2, 9)
if (2, 9) in trans_int:
    i = trans_int.index((2, 9))
    bax.annotate(
        None, xy=(w_trans[i], n_trans[i]), xytext=(4.84, 0.05),
        arrowprops=dict(facecolor=color_2, edgecolor=color_2,
                        arrowstyle="->", alpha=1,linewidth=linewidth_arrow))

# Drive arrows
bax.set_ylim(0, 0.8)
draw_drive_arrow(bax, wd1_phi, r"$\omega_{d1}$", dx_label=-0.09, dy=0.085, color='c')
draw_drive_arrow(bax, wd2_phi, r"$\omega_{d2}$", dx_label=-0.09, dy=0.085, color='c')

# Axes / save
bax.grid(linewidth=0.2, linestyle="--")
bax.set_xlabel(r"$\omega_{\rm i-j}\ (2\pi\cdot {\rm GHz})$", labelpad=10)
bax.set_ylabel(r"$n_\phi^{\rm i-j}$", labelpad=12)

fig.savefig("pdf_before_inkscape/xgate_nphi.pdf",dpi=300, bbox_inches="tight", pad_inches=0.02)
```
