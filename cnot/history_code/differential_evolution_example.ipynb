```python
import numpy as np
from scipy.optimize import differential_evolution, NonlinearConstraint

# Define the objective function
def objective_function(vars):
    x, y, z = vars
    return x**2 + y**2 + z**2  # A simple quadratic function to minimize

# Define the constraint function
def constraint(vars):
    x, y, z = vars
    return y - x  # Ensure that y - x >= 0 which implies y >= x

# Create the constraint
nonlinear_constraint = NonlinearConstraint(constraint, - np.inf, 0)

# Define bounds for the variables
bounds = [(2, 10.), (3, 10.), (4, 10.)]  # Both x and y are between 0 and 10

# Perform the differential evolution optimization
result = differential_evolution(objective_function, bounds, constraints=(nonlinear_constraint,))

# Print the result
print("Optimized variables:", result.x)
print("Function value:", result.fun)
```

```python
import numpy as np
from scipy.optimize import shgo

# Define the objective function
def objective_function(x):
    return np.sin(x[0]) + np.cos(x[1])

# Define the bounds for the variables
bounds = [(0, 2 * np.pi), (0, 2 * np.pi)]

# Set the options to display optimization progress
options = {'disp': True}

# Perform the SHGO optimization
result = shgo(objective_function, bounds, options=options)

# Print the result
print("Optimized variables:", result.x)
print("Function value:", result.fun)
```
